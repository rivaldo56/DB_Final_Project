```python
from transformers import MarianMTModel, MarianTokenizer

# English → Swahili
en_to_sw = "Helsinki-NLP/opus-mt-en-swc"
tokenizer_en2sw = MarianTokenizer.from_pretrained(en_to_sw)
model_en2sw = MarianMTModel.from_pretrained(en_to_sw)

# Swahili → English
sw_to_en = "Helsinki-NLP/opus-mt-swc-en"
tokenizer_sw2en = MarianTokenizer.from_pretrained(sw_to_en)
model_sw2en = MarianMTModel.from_pretrained(sw_to_en)
```

```python
text = "Education is the key to success."
inputs = tokenizer_en2sw(text, return_tensors="pt", padding=True)
translated = model_en2sw.generate(**inputs)
print(tokenizer_en2sw.decode(translated[0], skip_special_tokens=True))
```

```python
def translate_en_to_sw(text):
    inputs = tokenizer_en2sw(text, return_tensors="pt", padding=True)
    translated = model_en2sw.generate(**inputs)
    return tokenizer_en2sw.decode(translated[0], skip_special_tokens=True)

def translate_sw_to_en(text):
    inputs = tokenizer_sw2en(text, return_tensors="pt", padding=True)
    translated = model_sw2en.generate(**inputs)
    return tokenizer_sw2en.decode(translated[0], skip_special_tokens=True)

def bilingual_edu_bot(text):
    # Simple check if the sentence contains Swahili keywords
    swahili_words = ["ni", "kwa", "mimi", "wewe", "yeye", "hii", "huyo"]
    if any(word in text.lower().split() for word in swahili_words):
        print("🗣 Detected: Swahili")
        english = translate_sw_to_en(text)
        print("🧠 English translation:", english)
        answer = f"The topic '{english}' refers to an important concept in education."
        print("🤖 Bot:", answer)
        swahili_reply = translate_en_to_sw(answer)
        return f"💬 Swahili reply: {swahili_reply}"
    else:
        print("🗣 Detected: English")
        swahili = translate_en_to_sw(text)
        print("🧠 Swahili translation:", swahili)
        answer = f"{text} is an important concept for students to learn."
        swahili_reply = translate_en_to_sw(answer)
        return f"💬 Swahili reply: {swahili_reply}"

# Try a test
print(bilingual_edu_bot("Explain photosynthesis kwa Kiswahili"))
```

```python
samples = [
    "Education is the key to success.",
    "What is democracy?",
    "Nipe tafsiri ya neno 'computer'.",
    "Describe climate change in Swahili."
]

for s in samples:
    print(f"\n🧩 Input: {s}")
    print(bilingual_edu_bot(s))
    print("-" * 70)
```
